Etapa A

```python
import pandas as pd
import numpy as np
import urllib.request
import zipfile
import os
import glob

pd.set_option("display.max_columns", None)
```

Carregar dataset :

```python
url_dataset = "https://archive.ics.uci.edu/static/public/851/steel+industry+energy+consumption.zip"
caminho_zip = "steel_industry_energy_consumption.zip"
pasta_extraida = "steel_industry_energy_consumption"

urllib.request.urlretrieve(url_dataset, caminho_zip)

with zipfile.ZipFile(caminho_zip, "r") as arquivo_zip:
    arquivo_zip.extractall(pasta_extraida)

# Localiza o CSV dentro da pasta extraída (robusto a subpastas ou variação no nome do arquivo)
arquivos_csv = glob.glob(os.path.join(pasta_extraida, "**", "*.csv"), recursive=True)
print("Arquivo(s) CSV encontrado(s):", arquivos_csv)

caminho_csv = arquivos_csv[0]
```

```python
df_original = pd.read_csv(caminho_csv)

print("Formato do dataset original:", df_original.shape)
df_original.head()
```

```python
df_original.info()
```

Selecionar colunas

```python
colunas_selecionadas = [
    "Usage_kWh",
    "Lagging_Current_Reactive.Power_kVarh",
    "Leading_Current_Reactive_Power_kVarh",
    "Lagging_Current_Power_Factor",
    "Leading_Current_Power_Factor",
    "WeekStatus",
    "Day_of_week",
    "Load_Type",
]

df_selecionado = df_original[colunas_selecionadas].copy()
df_selecionado.head()
```

```python
print("Categorias em Load_Type:", df_selecionado["Load_Type"].unique())
print("Categorias em WeekStatus:", df_selecionado["WeekStatus"].unique())
```

```python
df_selecionado.isnull().sum()
```

```python
amostra_dataset2 = df_selecionado.sample(frac=0.20, random_state=42).reset_index(drop=True)

print("Formato da amostra (20%):", amostra_dataset2.shape)
amostra_dataset2.head()
```

```python
amostra_dataset2.to_csv("amostra_dataset2_steel.csv", index=False)
print("Amostra exportada em amostra_dataset2_steel.csv")
```

Etapa B:

```python
df = pd.read_csv("amostra_dataset2_steel.csv")

df = df.rename(columns={
    "Usage_kWh": "Consumo_kWh",
    "Lagging_Current_Power_Factor": "Fator_Potencia_Atrasado",
    "Leading_Current_Power_Factor": "Fator_Potencia_Adiantado",
})

df.columns.tolist()
```

```python
consumo_maximo = df["Consumo_kWh"].max()
limiar_75 = 0.75 * consumo_maximo

print(f"Maior consumo registrado na amostra: {consumo_maximo:.2f} kWh")
print(f"Limiar (75% do máximo): {limiar_75:.2f} kWh")
```

```python
df_alto_consumo = df[df["Consumo_kWh"] > limiar_75]

qtd_alto_consumo = len(df_alto_consumo)
pct_alto_consumo = qtd_alto_consumo / len(df) * 100

print(f"Registros com consumo elevado: {qtd_alto_consumo}")
print(f"Isso representa {pct_alto_consumo:.2f}% da amostra")
```

```python
rotulo_carga_maxima = [c for c in df["Load_Type"].unique() if "max" in c.lower()][0]
print("Rótulo identificado para carga máxima:", rotulo_carga_maxima)

qtd_carga_maxima = (df_alto_consumo["Load_Type"] == rotulo_carga_maxima).sum()
pct_carga_maxima_no_grupo = qtd_carga_maxima / qtd_alto_consumo * 100

pct_carga_maxima_geral = (df["Load_Type"] == rotulo_carga_maxima).mean() * 100

print(f"Registros de carga máxima dentro do grupo de alto consumo: {qtd_carga_maxima} ({pct_carga_maxima_no_grupo:.2f}%)")
print(f"Participação da carga máxima na amostra completa: {pct_carga_maxima_geral:.2f}%")
```

```python
df["Fator_Potencia_Atrasado"].describe()
```

```python
limite_fp_baixo = df["Fator_Potencia_Atrasado"].quantile(0.25)
print(f"Limite escolhido para fator de potência baixo: {limite_fp_baixo:.2f}")
```

```python
df_atencao = df[
    (df["Consumo_kWh"] > limiar_75) &
    (df["Fator_Potencia_Atrasado"] < limite_fp_baixo)
]

qtd_atencao = len(df_atencao)
pct_atencao_amostra = qtd_atencao / len(df) * 100
pct_atencao_no_grupo_alto = (qtd_atencao / qtd_alto_consumo * 100) if qtd_alto_consumo else 0

print(f"Registros com consumo elevado E fator de potência baixo: {qtd_atencao}")
print(f"Isso representa {pct_atencao_amostra:.2f}% da amostra total")
print(f"e {pct_atencao_no_grupo_alto:.2f}% do grupo que já tinha consumo elevado")

df_atencao.head()
```
