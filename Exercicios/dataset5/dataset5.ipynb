# Situação

Um operador de um portfólio de fontes renováveis deseja comparar a ocorrência de períodos de alta produção solar e alta produção eólica. Como as duas fontes possuem escalas próprias, cada uma deverá ser comparada com o seu próprio valor máximo.

```python
# Amostra gerada no Orange
dados = pd.read_csv("/content/amostra_producao_energia.csv")

print("--- Primeiros Registros ---")
print(dados.head())
```

# Etapa B — Python / Pandas

1. Renomeie as variáveis principais para Geracao_Solar e Geracao_Eolica.

2. Determine o valor máximo registrado para cada fonte.

3. Calcule separadamente 70% do máximo solar e 70% do máximo eólico.

4. Crie um DataFrame para registros de alta geração solar e outro para registros de alta geração eólica.

5. Conte os registros de cada DataFrame e calcule seus respectivos percentuais.

6. Compare qual fonte aparece com maior frequência acima de 70% do seu próprio máximo.

7. Explique por que não seria adequado utilizar diretamente o mesmo valor numérico de potência como limite para as duas fontes sem antes observar a escala de cada variável.

```python
# 1. Renomeie as variáveis principais para Geracao_Solar e Geracao_Eolica.
dados_renomeados = dados.copy()

# Criar colunas para Geracao_Solar e Geracao_Eolica baseadas na coluna 'Source'
dados_renomeados['Geracao_Solar'] = dados_renomeados.apply(
    lambda row: row['Production'] if row['Source'] == 'Solar' else 0, axis=1
)
dados_renomeados['Geracao_Eolica'] = dados_renomeados.apply(
    lambda row: row['Production'] if row['Source'] == 'Wind' else 0, axis=1
)

print("--- Primeiros Registros com Novas Colunas de Geração ---")
display(dados_renomeados.head())
```

```python
# 2. Determine o valor máximo registrado para cada fonte.
max_geracao_solar = dados_renomeados['Geracao_Solar'].max()
max_geracao_eolica = dados_renomeados['Geracao_Eolica'].max()

print(f"\nValor Máximo de Geração Solar: {max_geracao_solar}")
print(f"Valor Máximo de Geração Eólica: {max_geracao_eolica}")
```

```python
# 3. Calcule separadamente 70% do máximo solar e 70% do máximo eólico.
setenta_porcento_solar = max_geracao_solar * 0.70
setenta_porcento_eolica = max_geracao_eolica * 0.70

print(f"\n70% do Máximo Solar: {setenta_porcento_solar:.2f}")
print(f"70% do Máximo Eólico: {setenta_porcento_eolica:.2f}")
```

```python
# 4. Crie um DataFrame para registros de alta geração solar e outro para registros de alta geração eólica.
alta_geracao_solar_df = dados_renomeados[dados_renomeados['Geracao_Solar'] > setenta_porcento_solar].copy()
alta_geracao_eolica_df = dados_renomeados[dados_renomeados['Geracao_Eolica'] > setenta_porcento_eolica].copy()

print("\n--- Registros de Alta Geração Solar (Primeiros 5) ---")
display(alta_geracao_solar_df.head())
print(f"Total de registros de alta geração solar: {alta_geracao_solar_df.shape[0]}")

print("\n--- Registros de Alta Geração Eólica (Primeiros 5) ---")
display(alta_geracao_eolica_df.head())
print(f"Total de registros de alta geração eólica: {alta_geracao_eolica_df.shape[0]}")
```

```python
# 5. Conte os registros de cada DataFrame e calcule seus respectivos percentuais.

total_registros = dados_renomeados.shape[0]

count_alta_solar = alta_geracao_solar_df.shape[0]
percent_alta_solar = (count_alta_solar / total_registros) * 100

count_alta_eolica = alta_geracao_eolica_df.shape[0]
percent_alta_eolica = (count_alta_eolica / total_registros) * 100

print(f"\nNúmero de registros com alta geração solar: {count_alta_solar} ({percent_alta_solar:.2f}% do total)")
print(f"Número de registros com alta geração eólica: {count_alta_eolica} ({percent_alta_eolica:.2f}% do total)")
```

```python
# 6. Compare qual fonte aparece com maior frequência acima de 70% do seu próprio máximo.
if count_alta_solar > count_alta_eolica:
    print("\nA geração solar ocorre com maior frequência acima de 70% do seu próprio máximo.")
elif count_alta_eolica > count_alta_solar:
    print("\nA geração eólica ocorre com maior frequência acima de 70% do seu próprio máximo.")
else:
    print("\nA geração solar e eólica ocorrem com frequência semelhante acima de 70% do seu próprio máximo.")
```

## 7. Explique por que não seria adequado utilizar diretamente o mesmo valor numérico de potência como limite para as duas fontes sem antes observar a escala de cada variável.

Usar o mesmo valor numérico de potência como limite fixo para solar e eólica seria inadequado porque as duas fontes têm escalas de produção diferentes (por exemplo, o máximo eólico de 23446 supera o máximo solar de 16578) e perfis de geração intrínsecos distintos, ligados a fatores ambientais diferentes (radiação solar vs. velocidade do vento); assim, um limite absoluto representaria porcentagens muito diferentes do potencial máximo de cada fonte, distorcendo a comparação, enquanto uma métrica relativa — como 70% do próprio máximo de cada fonte — permite avaliar o quão perto cada uma está de sua capacidade individual, tornando a análise de "alta produção" justa e significativa, já que esse conceito depende do potencial específico de cada tecnologia e não de um ponto arbitrário numa escala comum.
