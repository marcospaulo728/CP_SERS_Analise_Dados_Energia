```python
import pandas as pd
```

```python
df = pd.read_csv("amostra_10pct.csv")

df.columns = ["potencia_ativa", "potencia_reativa", "tensao", "corrente", "sub1", "sub2", "sub3"]
df.head()
```

```python
maximo = df["potencia_ativa"].max()
print("Potencia ativa maxima:", maximo)
```

```python
limite = maximo * 0.75
print("Limite (75% do maximo):", limite)

acima = df[df["potencia_ativa"] > limite]
acima.head()
```

```python
quantidade = len(acima)
percentual = quantidade / len(df) * 100
print("Quantidade de registros:", quantidade)
print("Percentual:", round(percentual, 3), "%")
```

```python
corrente_media = df["corrente"].mean()
print("Corrente media da amostra:", corrente_media)
```

```python
acima_dois = df[(df["potencia_ativa"] > limite) & (df["corrente"] > corrente_media)]
print("Quantidade de registros:", len(acima_dois))
acima_dois.head()
```
