# Questão 3 — Power Consumption of Tetouan City

## Etapa B — Python / Pandas

**Objetivo:** identificar qual das três zonas apresenta o maior pico de consumo e observar o efeito da condição de temperatura nos registros de maior demanda.




## Carregamento do dataset

```python
import os
import pandas as pd

arquivo_csv = "Tetuan City power consumption.csv"

if not os.path.exists(arquivo_csv):
    from google.colab import files
    uploaded = files.upload()
    arquivo_csv = next(iter(uploaded.keys()))

dados = pd.read_csv(arquivo_csv)

print(f"Quantidade de registros: {len(dados)}")
display(dados.head())
```

## 1. Renomear as três variáveis de consumo

```python
dados = dados.rename(columns={
    "Zone 1 Power Consumption": "Consumo_Zona_1",
    "Zone 2  Power Consumption": "Consumo_Zona_2",
    "Zone 3  Power Consumption": "Consumo_Zona_3"
})

print("Colunas após a renomeação:")
print(dados.columns.tolist())
```

## 2. Determinar o consumo máximo registrado em cada zona

```python
maximos = dados[
    ["Consumo_Zona_1", "Consumo_Zona_2", "Consumo_Zona_3"]
].max()

print("Consumo máximo por zona:")
print(maximos)
```

### Resultado dos máximos

- **Zona 1:** 52204.39512
- **Zona 2:** 37408.86076
- **Zona 3:** 47598.32636


## 3. Identificar qual zona apresenta o maior pico de consumo

```python
zona_maior = maximos.idxmax()
maximo_maior = maximos.max()

print(f"Zona com maior pico: {zona_maior}")
print(f"Maior pico de consumo: {maximo_maior:.5f}")
```

### Resultado

A **Zona 1** apresenta o maior pico de consumo, com valor máximo de **52204.39512**.


## 4. Calcular 70% do valor máximo e criar o DataFrame de alto consumo

```python
limiar_70 = 0.70 * maximo_maior

df_alto = dados[dados[zona_maior] > limiar_70].copy()

print(f"70% do máximo da Zona 1: {limiar_70:.5f}")
print(f"Registros acima do limiar: {len(df_alto)}")
display(df_alto.head())
```

### Resultado

O limiar correspondente a **70% do máximo da Zona 1** é **36543.07658**.

O DataFrame `df_alto` reúne todos os registros em que o consumo da Zona 1 é superior a esse valor.


## 5. Contar os registros e calcular o percentual

```python
qtd_alto = len(df_alto)
perc_alto = qtd_alto / len(dados) * 100

print(f"Total de registros do dataset: {len(dados)}")
print(f"Registros acima de 70% do máximo: {qtd_alto}")
print(f"Percentual: {perc_alto:.2f}%")
```

### Resultado

Foram encontrados **14718 registros** com consumo da Zona 1 acima de 70% do valor máximo.

Esses registros correspondem a **28.08%** dos **52416 registros** do dataset.


## 6. Calcular a temperatura média e criar um segundo DataFrame

O segundo DataFrame utiliza simultaneamente as condições:

1. consumo da Zona 1 acima de 70% do máximo;
2. temperatura acima da média do dataset.


```python
temp_media = dados["Temperature"].mean()

df_alto_temp = dados[
    (dados["Consumo_Zona_1"] > limiar_70) &
    (dados["Temperature"] > temp_media)
].copy()

qtd_alto_temp = len(df_alto_temp)
perc_alto_temp = qtd_alto_temp / len(dados) * 100

print(f"Temperatura média: {temp_media:.2f}")
print(f"Registros que atendem às duas condições: {qtd_alto_temp}")
print(f"Percentual sobre o dataset: {perc_alto_temp:.2f}%")

display(df_alto_temp.head())
```

### Resultado

A temperatura média registrada é de aproximadamente **18.81 °C**.

Ao exigir simultaneamente consumo elevado e temperatura acima da média, restaram **9917 registros**, equivalentes a **18.92%** do dataset.


## 7. Comparar os dois conjuntos

```python
reducao = qtd_alto - qtd_alto_temp
reducao_relativa = reducao / qtd_alto * 100

comparacao = pd.DataFrame({
    "Conjunto": [
        "Consumo acima de 70% do máximo",
        "Consumo acima de 70% + temperatura acima da média"
    ],
    "Quantidade": [qtd_alto, qtd_alto_temp],
    "Percentual_do_dataset": [perc_alto, perc_alto_temp]
})

display(
    comparacao.style.format({
        "Percentual_do_dataset": "{:.2f}%"
    })
)

print(f"Redução na quantidade de registros: {reducao}")
print(f"Redução em relação ao primeiro DataFrame: {reducao_relativa:.2f}%")
```

### Interpretação

O primeiro DataFrame contém **14718 registros (28.08%)** de alto consumo da Zona 1.

Após incluir a condição de **temperatura acima da média**, o segundo DataFrame passou a conter **9917 registros (18.92%)**.

Portanto, houve uma redução de **4801 registros**, o que corresponde a aproximadamente **32.62%** dos registros que inicialmente apresentavam consumo elevado.

Isso ocorre porque o segundo filtro é mais restritivo: para permanecer no novo conjunto, o registro precisa atender **às duas condições ao mesmo tempo**. Assim, parte dos momentos de alto consumo ocorreu com temperatura igual ou abaixo da média e foi excluída do segundo DataFrame.

### Conclusão

A **Zona 1** foi a zona com o maior pico de consumo, atingindo **52204.39512**. O limite de 70% desse pico foi **36543.07658**. Foram identificados **14718 registros (28.08%)** acima desse limite.

Quando também foi exigida temperatura superior à média de **18.81 °C**, a quantidade caiu para **9917 registros (18.92%)**. Portanto, a inclusão da condição ambiental reduziu o conjunto analisado, mostrando que nem todos os períodos de maior demanda da Zona 1 aconteceram sob temperaturas acima da média.
